```python
import wfdb
import matplotlib.pyplot as plt
```

```python
# Load the ECG signal for record 100.
record = wfdb.rdrecord('../data/raw/100')

# Load the beat annotations using the atr annotation file.
annotation = wfdb.rdann('../data/raw/100', 'atr')

# Print the signal dimensions and number of annotated beats.
print(f'Signal shape: {record.p_signal.shape}')
print(f'Annotated beats: {len(annotation.sample)}')
```

```python
wfdb.plot_wfdb(
    record=record,
    annotation=annotation,
    title='Record 100',
    time_units='seconds'
)
```
